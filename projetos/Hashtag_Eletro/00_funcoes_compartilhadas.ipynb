```python
import pandas as pd
import numpy as np
from pathlib import Path



def eliminar_linhas_colunas_em_branco(df):
    df_linhas_limpas = df.dropna(how="all")   # remove linhas onde todas as células estão em branco/nulos
    df_colunas_limpas = df_linhas_limpas.dropna(how="all", axis=1)  # Remove as colunas onde todas as células estão em branco/nulos
    return df_colunas_limpas


def padronizar_colunas_texto(coluna: pd.Series):
    coluna_tratada = coluna.astype(str).str.strip().str.lower().replace("", np.nan)
    return coluna_tratada

def padronizar_textos_tabela(df: pd.DataFrame):
    colunas_texto = df.select_dtypes(include=["object"]).columns
    
    for coluna in colunas_texto:
        df[coluna] = padronizar_colunas_texto(df[coluna])

    return df
```
