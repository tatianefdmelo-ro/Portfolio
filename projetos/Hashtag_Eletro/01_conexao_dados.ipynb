```python
import pandas as pd
from pathlib import Path
import openpyxl

dir_origem = Path.cwd() / "dados"
dir_destino = dir_origem / "bases_limpas"

# caso a pasta de destino não exista, será criada
dir_destino.mkdir(parents=True, exist_ok=True)
```

```python
def carregar_base(nome_base: str, diretorio: Path = dir_origem, skiprows=0)-> pd.DataFrame:
    caminho = diretorio / nome_base
    extensao = caminho.suffix.lower()
    
    # Detalhes de I/O isolados por formato
    try: 
        if extensao == '.csv':
            return pd.read_csv(caminho, sep=';', encoding='utf-8')
        elif extensao in ['.xlsx', '.xls']:
            return pd.read_excel(caminho, skiprows=skiprows)
        elif extensao == '.json':
            return pd.read_json(caminho)
        else:
            raise ValueError(f"Extensão {extensao} não suportada.")
    except Exception as e:
        raise e
```

```python
def carregar_a_partir_de_arquivos_separados(nome_base: str)-> pd.DataFrame:
    # Procura por qualquer arquivo que comece com o nome_base (ex: 'vendas'-> [vendas_2020.xlsx, vendas_2021.xlsx, 2023_vendas.xlsx])
    arquivos = list(dir_origem.glob(f"*{nome_base}*"))

    
    if not arquivos:
        print(f"Falha função 'carregar_a_partir_de_arquivos_separados'. ARGS: {nome_base} ")
        raise Exception("Nenhum arquivo e encontrado.")

    dfs = [] # 4 dataframes panda

    for base in arquivos:
        dfs.append(carregar_base(base.name))

    df = pd.concat(dfs, ignore_index=True)
    
    return df
```

```python
def carregar_a_partir_de_planilhas_de_um_excel(nome_arquivo_excel: str)-> pd.DataFrame:
    caminho = dir_origem / nome_arquivo_excel
    
    if not caminho.exists():
        print(f"Arquivo Excel {nome_arquivo_excel} não encontrado.")
        raise Exception(f"Arquivo {nome_arquivo_excel} não encontrado.")
    
    # sheet_name=None retorna um dicionário: {nome_da_aba: DataFrame}
    df = pd.read_excel(caminho, sheet_name=None)

    lista_de_planilhas_tabelas = df.values() # Cria uma lista de dataframes
    
    return pd.concat(lista_de_planilhas_tabelas, ignore_index=True)


```

```python
def salvar_base(df: pd.DataFrame, nome_arquivo: str)-> None:
    caminho_final = dir_destino / f"{nome_arquivo}.csv"
    df.to_csv(caminho_final, index=False, sep=";", encoding="utf-8")

    print(f"Base salva com sucesso em: {caminho_final}") 
    return
    

# def salvar_base_em_excel(df: pd.DataFrame, nome_arquivo: str)-> None:
#     caminho_final = dir_destino / nome_arquivo / ".xlsx"
#     df.to_excel(caminho_final, index=False, encoding='utf-8', engine=openpyxl)
#     print(f"Base salva com sucesso em: {caminho_final}")
#     return


    
```
